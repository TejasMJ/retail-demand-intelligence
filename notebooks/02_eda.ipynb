#### Importing the libraries

##### Time-Based Analysis

##### Daily Demand

```python
# Calculating total Demand for each date.

daily_demand = (
    df.groupby('Date', as_index=False)['Demand']
    .sum()
)

# Plotting daily Demand to identify changes in demand over time.

fig = px.line(
    daily_demand,
    x='Date',
    y='Demand',
    title='Daily Demand Over Time'
)

fig.show()
```

##### Monthly Demand

```python
# Creating a temporary Month column for monthly demand analysis.
df['Month'] = df['Date'].dt.to_period('M').astype(str)

# Calculating total Demand for each month.
monthly_demand = (
    df.groupby('Month', as_index=False)['Demand']
    .sum()
)

# Plotting monthly Demand to identify broader demand patterns.
fig = px.line(
    monthly_demand,
    x='Month',
    y='Demand',
    title='Monthly Demand Trend'
)

fig.update_xaxes(tickangle=45)

fig.show()
```

#### Outlier Analysis

```python
# Calculating the first and third quartiles for the numerical columns.

Q1 = df[numerical_columns].quantile(0.25)

Q3 = df[numerical_columns].quantile(0.75)
```

```python
# Calculating the interquartile range for the numerical columns.

IQR = Q3 - Q1
```

```python
# Identifying potential outliers using the IQR method.

outlier_summary = {}

for col in numerical_columns:

    lower_bound = Q1[col] - 1.5 * IQR[col]
    upper_bound = Q3[col] + 1.5 * IQR[col]

    outlier_count = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_summary[col] = outlier_count
```

```python
# Creating a summary of potential outliers across the numerical columns.

outlier_summary = (
    pd.DataFrame(
        outlier_summary.items(),
        columns=['Column', 'Outlier Count']
    )
    .sort_values('Outlier Count', ascending=False)
)

outlier_summary
```

```python
# Plotting the number of potential outliers for each numerical column.

fig = px.bar(
    outlier_summary,
    x='Column',
    y='Outlier Count',
    title='Potential Outliers Across Numerical Columns'
)

fig.show()
```
