#### Based on the dataset documentation, the meaning of each column in the Retail Store Inventory and Demand Forecasting dataset is as follows:

- Date - Date on which the store-level product observation was recorded.
- Store ID - Unique identifier assigned to each retail store.
- Product ID - Unique identifier assigned to each product.
- Category - Product category to which the item belongs.
- Region - Geographic region associated with the store.
- Inventory Level - Number of units of the product available in inventory on the given date.
- Units Sold - Number of units of the product sold on the given date.
- Units Ordered - Number of units of the product ordered for replenishment on the given date.
- Price - Selling price of the product on the given date.
- Discount - Discount applied to the product on the given date.
- Weather Condition - Weather condition recorded for the store/location on the given date.
- Promotion - Indicates whether the product was subject to a promotional campaign or activity.
- Competitor Pricing - Price of the comparable product offered by competitors.
- Seasonality - Seasonal factor or period associated with the observation that may influence demand.
- Epidemic - Indicator representing whether an epidemic-related condition was present during the observation period.
- Demand - Estimated or observed demand for the product, representing the quantity demanded and serving as the primary target variable for demand prediction.

#### Checking for missing values per column (including empty strings)

```python
# Count of NaN / None values per column
nan_count = df.isnull().sum()

# Count of empty strings per column
empty_count = (df == '').sum()

# Total missing values (NaN + empty strings)
total_missing = nan_count + empty_count

# Percentage of missing values
missing_percentage = (total_missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': total_missing,
    'Percentage (%)': missing_percentage
}, index=df.columns).sort_values(by='Percentage (%)', ascending=False)

print(missing_df)
```

#### Observations on Missing Values

The dataset contains no missing values across any of the 16 columns, indicating that the dataset is complete and does not require any missing-value treatment at this stage.
