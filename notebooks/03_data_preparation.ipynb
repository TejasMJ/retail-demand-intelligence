#### Importing the libraries.

```python
import pandas as pd
import numpy as np
```

#### Loading the dataset

```python
df = pd.read_csv('../data/raw/sales_data.csv')
```

```python
df.head()
```

#### Creating a Copy of the Dataset

```python
# Creating a copy of the original dataset for data preparation.

df_prepared = df.copy()
```

#### Converting the Date Column

```python
# Converting the Date column from object to datetime format.

df_prepared['Date'] = pd.to_datetime(
    df_prepared['Date']
)
```

```python
# Checking the updated data types.

df_prepared.dtypes
```

#### Checking for Duplicate Records

```python
# Even though the dataset may not contain duplicates, doule checking to validate this before modeling.

# Checking the number of duplicate rows in the dataset.

duplicate_count = df_prepared.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

#### Validating Missing Values

```python
# Checking for missing values before preparing the dataset.

missing_values = df_prepared.isnull().sum()

print(missing_values)
```

#### Checking for Invalid Negative Values

```python
# Defining the numerical columns for validation.

numerical_columns = [
    'Inventory Level',
    'Units Sold',
    'Units Ordered',
    'Price',
    'Discount',
    'Competitor Pricing',
    'Demand'
]


# Checking for negative values in the numerical columns.

for column in numerical_columns:
    
    negative_values = (
        df_prepared[column] < 0
    ).sum()
    
    print(
        f"{column}: {negative_values} negative values"
    )
```

#### Validating Binary Columns

```python
# Defining the binary columns for validation.

binary_columns = [
    'Promotion',
    'Epidemic'
]

# Checking the unique values in the binary columns.

for column in binary_columns:
    
    print(f"\n{column}:")
    print(df_prepared[column].unique())
```

#### Converting Categorical Columns

```python
# Defining the categorical columns.

categorical_columns = [
    'Store ID',
    'Product ID',
    'Category',
    'Region',
    'Weather Condition',
    'Seasonality'
]

# Converting the categorical columns to category data type.

for column in categorical_columns:
    
    df_prepared[column] = (
        df_prepared[column].astype('category')
    )
```

```python
# Checking the updated data types after conversion.

df_prepared.dtypes
```

#### Sorting the Dataset by Date

```python
# Sorting the dataset chronologically by date.

df_prepared = df_prepared.sort_values(
    by='Date'
).reset_index(
    drop=True
)
```

```python
# Displaying the first few records after sorting by date.

df_prepared.head()
```

#### Final Data Validation

```python
# Checking the final dataset information.

df_prepared.info()
```

```python
# Checking for missing values in the prepared dataset.

df_prepared.isnull().sum()
```

```python
# Checking the final dataset shape.

print(
    "Final Dataset Shape:",
    df_prepared.shape
)
```

#### Saving the Prepared Dataset

```python
# Saving the prepared dataset for the next stage of the project.

df_prepared.to_csv(
    '../data/processed/prepared_data.csv',
    index=False
)
```
